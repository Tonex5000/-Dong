## Import Libraries

```python
#Import General libraries
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt

#Balancing the dataset
from imblearn.over_sampling import SMOTE


#Import Sklearn libraries for learning
from sklearn import metrics, linear_model, ensemble
from sklearn.preprocessing import StandardScaler
import xgboost as xgb
import lightgbm as lgbm
from sklearn.linear_model import LogisticRegression
from xgboost import XGBClassifier
from lightgbm import LGBMClassifier
from sklearn.feature_selection import SelectKBest, chi2, f_classif, mutual_info_classif
from sklearn.ensemble import GradientBoostingClassifier
from sklearn.model_selection import train_test_split,cross_val_score, RepeatedStratifiedKFold, GridSearchCV,ShuffleSplit,cross_validate

# SKlearn libraries for evaluation.
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix, roc_auc_score


#Import Joblib to save model for production
import joblib
```

## Part I: Read Data and EDA

## Cleaning data

```python
#Checking for duplicate
df.duplicated().sum()
```

```python
#Checking for NaN(Not A Number)
df.isnull().sum()
```

```python
invalid_zero_cols = ['Glucose', 'BloodPressure', 'SkinThickness', 'Insulin', 'BMI']
for col in invalid_zero_cols:
    df[col] = df[col].replace(0, np.nan)
    df[col] = df[col].fillna(df[col].median())
```

```python
df.hist(bins=50, figsize=(20,15), edgecolor='black')
```

```python
df['Pregnancies'] = np.log1p(df['Pregnancies'])
df['Insulin'] = np.log1p(df['Insulin'])
df['DiabetesPedigreeFunction'] = np.log1p(df['DiabetesPedigreeFunction'])
df['Age'] = np.sqrt(df['Age'])
```



## Part II: Feature selection and feature importance

```python
plt.figure(figsize=(12, 6))
sns.heatmap(df.corr(), annot=True, cmap='coolwarm')
plt.title('Correlation Heatmap')
plt.show()
```

```python
X = df.drop('Outcome', axis=1)
y = df['Outcome']
```

```python
selector = SelectKBest(score_func=f_classif, k='all')
X_new = selector.fit_transform(X, y)
```

```python
importance = selector.scores_
importance_series = (pd.Series(importance, index=X.columns)).sort_values(ascending=False)
print(importance_series)
```

```python
plt.figure(figsize=(10,6))
sns.barplot(x=importance_series.values, y=importance_series.index)
plt.xlabel('Score')
plt.ylabel('Features')
plt.title('Feature Importance')
plt.show()
```

## Part III: Model selection and data training

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42, stratify=y)
```

```python
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)
```

```python
model = LogisticRegression(max_iter=1000, random_state=42)
```

```python
cv = RepeatedStratifiedKFold(n_splits=5, n_repeats=3, random_state=42)
```

```python
scores = cross_val_score(model, X_train_scaled, y_train, scoring='accuracy', cv=cv)
scores
```

```python
mean_accuracy = scores.mean()
std_accuracy = scores.std()
print("Mean Accuracy:", mean_accuracy)
print("Standard Deviation of Accuracy:", std_accuracy)
print("Cross-validation scores:", scores)
```

```python
plt.figure(figsize=(8,5))
plt.plot(range(1, len(scores)+1), scores, marker='o', linestyle='-', color='b')
plt.title('Cross-Validation Scores')
plt.xlabel('Fold')
plt.ylabel('Accuracy')
plt.grid(True)
plt.show()
```

## Explore and search for the best model

```python
hyper_params_lgbm = {
    "objective": "binary",
    "boosting_type": "gbdt",
    "n_estimators": 500,
    "verbosity": -1
}


hyper_params_xgb = {
    "objective": "binary:logistic",
    "n_estimators": 500,
    "eval_metric": "logloss"
}
```

```python
cv_split= ShuffleSplit(n_splits=5, test_size=0.3, random_state=42)
```

```python
algorithm_list = {
    'LogisticReg': linear_model.LogisticRegression(max_iter=1000),
    'RandomForest': ensemble.RandomForestClassifier(n_estimators=200, random_state=42),
    'GradientBoost': ensemble.GradientBoostingClassifier(n_estimators=200, random_state=42),
    'XGBoost': xgb.XGBClassifier(**hyper_params_xgb),
    'LGBM': lgbm.LGBMClassifier(**hyper_params_lgbm)
}
```

```python
for key, model in algorithm_list.items():
  try:
    if key in 'LightGBM':
      model.fit(X_train, y_train, eval_set=[(X_test, y_test)], early_stopping_rounds=100, verbose=False)
    elif key in 'XGBoost':
      model.fit(X_train, y_train, eval_set=[(X_test, y_test)], verbose=False)
    else:
      model.fit(X_train, y_train)
    cv_results = cross_validate(model, X_train, y_train, cv=cv_split)
    y_pred = model.predict(X_test)

    acc_train = model.score(X_train, y_train)
    acc_test = model.score(X_test, y_test)

    cv_mean = cv_results['test_score'].mean()
    cv_std = cv_results['test_score'].std()

    precision = metrics.precision_score(y_test, y_pred)
    recall = metrics.recall_score(y_test, y_pred)
    f1 = metrics.f1_score(y_test, y_pred)
    roc_auc = metrics.roc_auc_score(y_test, model.predict_proba(X_test)[:, 1])

    print(f'Model - {key: <12} || Train Acc: {acc_train:.3f} | Test Acc: {acc_test:.3f} | CV Mean: {cv_mean:.3f} | CV STD: {cv_std:.3f}')
    print(f'Precision: {precision:.3f}, Recall: {recall:.3f}, F1: {f1:.3f}, ROC-AUC: {roc_auc:.3f}')
    print('----------------------------------------------------------------------------------------------------------')

  except Exception as e:
    print(f"Exception in {key}: {str(e)}")
    pass
```

```python
algorithm_list = {
    'LogisticReg': linear_model.LogisticRegression(max_iter=1000, class_weight='balanced'),
    'RandomForest': ensemble.RandomForestClassifier(n_estimators=200, class_weight='balanced', random_state=42),
    'GradientBoost': ensemble.GradientBoostingClassifier(n_estimators=200, random_state=42),
    'XGBoost': xgb.XGBClassifier(scale_pos_weight=500/268, random_state=42),  # weighting
    'LGBM': lgbm.LGBMClassifier(scale_pos_weight=500/268, random_state=42)
}
```

```python
smote = SMOTE(random_state=42)
X_train_bal, y_train_bal = smote.fit_resample(X_train, y_train)
```

```python
for key, model in algorithm_list.items():
    try:
        print(f"Training {key} ...")

        # Fit model differently depending on type
        if key =='LGBM':
            model.fit(
                X_train_bal, y_train_bal,
                eval_set=[(X_test, y_test)]
            )

        elif key == 'XGBoost':
            model.fit(
                X_train_bal, y_train_bal,
                eval_set=[(X_test, y_test)],
                verbose=False
            )

        else:
            model.fit(X_train_bal, y_train_bal)

        # Cross-validation on the original (unbalanced) X_train for fair scoring
        cv_results = cross_validate(model, X_train, y_train, cv=cv_split)

        # Predictions
        y_pred = model.predict(X_test)
        y_proba = model.predict_proba(X_test)[:, 1]

        # Scores
        acc_train = model.score(X_train_bal, y_train_bal)
        acc_test = model.score(X_test, y_test)
        cv_mean = cv_results['test_score'].mean()
        cv_std = cv_results['test_score'].std()

        precision = metrics.precision_score(y_test, y_pred)
        recall = metrics.recall_score(y_test, y_pred)
        f1 = metrics.f1_score(y_test, y_pred)
        roc_auc = metrics.roc_auc_score(y_test, y_proba)

        print(f'Model - {key: <12} || Train Acc: {acc_train:.3f} | Test Acc: {acc_test:.3f} | CV Mean: {cv_mean:.3f} | CV STD: {cv_std:.3f}')
        print(f'Precision: {precision:.3f}, Recall: {recall:.3f}, F1: {f1:.3f}, ROC-AUC: {roc_auc:.3f}')
        print('-' * 110)

    except Exception as e:
        print(f"Exception in {key}: {str(e)}")
```
